```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("../data/raw/music_data_raw.csv")
```

```python
targets = [
    "digit_span",
    "spatial_span",
    "melody_span"
]
```

```python
demographics = [
    "age",
    "sex",
    "years_of_education",
    "country",
    "ses_nuclear_family"
]
```

```python
music = [
    "gold_msi_musical_training",
    "gold_msi_general_sophistication",
    "gold_msi_active_engagement"
]
```

```python
personality = [
    "bfi_extraversion",
    "bfi_agreeableness",
    "bfi_conscientiousness",
    "bfi_negative_emotionality",
    "bfi_open_mindedness"
]
```

```python
selected_columns = (
    demographics +
    music +
    personality +
    targets
)

df_model = df[selected_columns]
```

```python
df_model.isnull().sum()
```

```python
missing_percentage = (
    df_model.isnull().mean() * 100
).round(2)

missing_percentage
```

```python
df_model = df_model.dropna()
```

```python
df_model.shape
```

```python
df_model.to_csv(
    "../data/processed/music_memory_processed.csv",
    index=False
)
```
